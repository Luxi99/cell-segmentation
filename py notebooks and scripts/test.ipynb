```python
import jwt
import time

private_key = open("/home/samu/ed25519-private.pem").read()

payload = {
    "sub": "2B2F72D2M7",
    "iat": int(time.time()),
    "exp": int(time.time()) + 86000
}

headers = {
    "alg": "EdDSA",
    "kid": "C5B8UBYGVX"
}

token = jwt.encode(
    payload,
    private_key,
    algorithm="EdDSA",
    headers=headers
)

print(token)
```
