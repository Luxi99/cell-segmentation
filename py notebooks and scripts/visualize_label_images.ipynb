```python
import os
import time
import numpy as np
import matplotlib.pyplot as plt
from skimage import io
from skimage.color import label2rgb

folder = "/home/samu/Documents/Laurea/Data/input/OSCC_PATIENTS/SOSPENSIONE/PR2/PR_2_Projet/exports"
files = sorted([f for f in os.listdir(folder) if f.endswith(".tif")])


for f in files:
    path = os.path.join(folder, f)
    
    labels = io.imread(path)
    colored = label2rgb(labels, bg_label=0)
    
    plt.figure(figsize=(8,8))
    plt.imshow(colored)
    plt.title(f)
    plt.axis('off')
    plt.show()
    
    time.sleep(2)
```
